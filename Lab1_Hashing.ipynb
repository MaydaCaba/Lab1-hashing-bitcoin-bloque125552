BLOQUE 125552

```python
import hashlib
import struct

# datos del bloque 125552
version=1
bloque_anterior = "00000000000008a3a41b85b8b29ad444def299fee21793cd8b9e567eab02cd81"
merkle_root = "2b12fcf1b09288fcaff797d71e950e71ae42b91e8bdb2304758dfcffc2b620e3"
tiempo = 1305998791
bits = 0x1a44b9f2
nonce = 0x9546a142

#convertiendo los numeros a little-endian
version_inv = struct.pack("<I", version)
tiempo_inv = struct.pack("<I", tiempo)
bits_inv = struct.pack("<I", bits)
nonce_inv = struct.pack("<I", nonce)

#volteamos los hashes al reves con [::-1]
bloque_anterior_inv = bytes.fromhex(bloque_anterior)[::-1]
merkle_root_inv = bytes.fromhex(merkle_root)[::-1]

#unimos los datos para formar la cabecera de 80 bytes
cabecera = (
    version_inv +
    bloque_anterior_inv +
    merkle_root_inv +
    tiempo_inv +
    bits_inv +
    nonce_inv
)

#aplicamos doble SHA-256
primer_hash = hashlib.sha256(cabecera).digest()
segundo_hash = hashlib.sha256(primer_hash).digest()

#volteamos el resultado final para leer
hash_final = segundo_hash[::-1].hex()
hash_oficial = "00000000000000001e8d6829a8a21adc5d38d0a473b144b6765798e61f98bd1d"

#imprimimos el hash resultante
print("tamaño cabecera:", len(cabecera))
print("Hex cabecera   :", cabecera.hex())
print("Hash calculado:", hash_final)
print("Hash oficial:", hash_oficial)
print("¿Son iguales?:", hash_final == hash_oficial)
```

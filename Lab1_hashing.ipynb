```python
import hashlib, struct

def sha256_texto(texto: str) -> str:
  datos = texto.encode("utf-8")
  return hashlib.sha256(datos).hexdigest()


def doble_sha256(datos_binarios: bytes) -> bytes:
    """Aplica SHA-256 dos veces (estándar del protocolo Bitcoin)."""
    primera_ronda = hashlib.sha256(datos_binarios).digest()
    segunda_ronda = hashlib.sha256(primera_ronda).digest()
    return segunda_ronda


def parte_1_propiedades_hash():
    print(sha256_texto("=" * 65))
    print(sha256_texto("PARTE 1: PROPIEDADES DE HASHING Y EFECTO AVALANCHA (SHA-256)"))
    print(sha256_texto("=" * 65))

    mensaje_original = "Bloque 125552 - Laboratorio1 Blockchain"
    # Ahora sí alteramos un solo carácter (la 'B' por 'b') para ver el efecto avalancha
    mensaje_alterado = "Bloque 125552 - Laboratorio1 blockchain"

    hash_1 = sha256_texto(mensaje_original)
    hash_2 = sha256_texto(mensaje_alterado)

    print(f"Texto 1 : {mensaje_original}")
    print(f"SHA-256 : {hash_1}\n")
    print(f"Texto 2 : {mensaje_alterado}")
    print(f"SHA-256 : {hash_2}\n")
    print(f"¿Son iguales?: {hash_1 == hash_2} (Efecto avalancha comprobado al ser diferentes)\n")


def parte_2_bloque_125552():
    print(sha256_texto("=" * 65))
    print(sha256_texto("PARTE 2: REPRODUCCIÓN DEL HASH DEL BLOQUE 125552 DE BITCOIN"))
    print(sha256_texto("=" * 65))

    # Datos reales del Bloque #125552 en formato humano (Big-Endian / enteros)
    version = 1
    prev_block_be = "00000000000008a3a41b85b8b29ad444def299fee21793cd8b9e567eab02cd81"
    merkle_root_be = "2b12fc1b072888f97be7d1e950e71ae42b91e8b3db2304758dfcffc2b620e3"
    # Nota: verificamos los 64 caracteres exactos del Merkle Root oficial del bloque 125552:
    merkle_root_be = "2b12fc1b072888f97be7d1e950e71ae42b91e8b3db2304758dfcffc2b620e3"
    merkle_root_le_hex = "e320b6c2fffc8d750423db8b1eb942ae710e951ed797f7affc8892b0f1fc122b"
    merkle_root_be = bytes.fromhex(merkle_root_le_hex)[::-1].hex()

    timestamp = 1305998791   # 2011-05-21 17:26:31 UTC (0x4dd7f5c7)
    bits = 0x1a44b9f2        # Objetivo de dificultad compacto (440711666)
    nonce = 2504433986       # Nonce ganador (0x9546a142)

    # 1. Serialización de los 6 campos a Little-Endian (total: 80 bytes)
    version_le = struct.pack("<I", version)               # 4 bytes
    prev_block_le = bytes.fromhex(prev_block_be)[::-1]    # 32 bytes invertidos
    merkle_root_le = bytes.fromhex(merkle_root_be)[::-1]  # 32 bytes invertidos
    timestamp_le = struct.pack("<I", timestamp)           # 4 bytes
    bits_le = struct.pack("<I", bits)                     # 4 bytes
    nonce_le = struct.pack("<I", nonce)                   # 4 bytes

    # 2. Concatenación de la cabecera completa (4 + 32 + 32 + 4 + 4 + 4 = 80 bytes)
    block_header = (
        version_le
        + prev_block_le
        + merkle_root_le
        + timestamp_le
        + bits_le
        + nonce_le
    )

    print(f"Longitud de cabecera : {len(block_header)} bytes")
    print(f"Cabecera Hex (LE)    : {block_header.hex()}\n")

    # 3. Cálculo de Doble SHA-256
    hash_binario_le = doble_sha256(block_header)

    # 4. Inversión de bytes a Big-Endian para visualización estándar
    hash_final_be = hash_binario_le[::-1].hex()
    hash_esperado = "00000000000000001e8d6829a8a21adc5d38d0a473b144b6765798e61f98bd1d"

    print(f"Hash en Little-Endian: {hash_binario_le.hex()}")
    print(f"Hash Calculado (BE)  : {hash_final_be}")
    print(f"Hash Oficial Bitcoin : {hash_esperado}")
    print(f"¿Coincide exactamente?: {hash_final_be == hash_esperado}")


if __name__ == "__main__":
    parte_1_propiedades_hash()
    parte_2_bloque_125552()
```
